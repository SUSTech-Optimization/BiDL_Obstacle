# Training and forward-solve curves

Run all cells from the repository root. The curves use the CSV files in `data/paper_curves/` and are saved to `outputs/figures/`.


```python
from pathlib import Path
import os

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

figs_dir = Path("outputs/figures")
figs_dir.mkdir(parents=True, exist_ok=True)
```

## Example 1: Bi-active set


```python
sampling_interval = 1
norm_csv_path = "data/paper_curves/example1/norm.csv"


df = pd.read_csv(norm_csv_path)
labels = {
    'rel_l2_error_y': 'State y',
    'rel_l2_error_v': 'Control u'
}

fig, ax = plt.subplots(figsize=(10,6))

# Plot curves with subsampling controlled by sampling_interval
lines = {}
epochs = df['epoch'].values
for key, name in labels.items():
    errs = df[key].values
    sampled_epochs = epochs[::sampling_interval]
    sampled_errs   = errs[::sampling_interval]
    line, = ax.semilogy(
        sampled_epochs,
        sampled_errs,
        lw=2,
        label=name
    )
    lines[key] = line

# Annotate the final errors using the curve colors
x_last = epochs[-1]
x0, x1 = ax.get_xlim()
x_text = x1 + 0.02*(x1-x0)
for key, name in labels.items():
    y_last = df[key].iloc[-1]
    color = lines[key].get_color()
    y_text = y_last * (0.7 if key=='rel_l2_error_y' else 1.7)
    mantissa, exponent = f"{y_last:.2e}".split('e')
    scientific_notation = f"{name}: ${mantissa} \\times 10^{{{int(exponent)}}}$"
    ax.annotate(
        scientific_notation,
        xy=(x_last, y_last),
        xytext=(x_text, y_text),
        textcoords='data',
        va='center', ha='left',
        fontsize=14,
        color=color,
        bbox=dict(boxstyle='round,pad=0.3', fc='white', ec=color, alpha=0.8),
        arrowprops=dict(arrowstyle='->', color=color, lw=1)
    )

ax.set_xlabel("Epoch", fontsize=16)
ax.set_ylabel(r"Relative $L^2$-error", fontsize=16)
ax.tick_params(axis='both', labelsize=14)
ax.legend(fontsize=18)
ax.set_xlim(0, x_text + 0.02*(x1-x0))

plt.tight_layout()
plt.savefig(os.path.join(figs_dir, 'example1_Bi_process.pdf'), bbox_inches='tight')
plt.show()
```

```python
sampling_interval = 1
forward_csv_path = "data/paper_curves/example1/forward_l2errors.csv"


df = pd.read_csv(forward_csv_path)
labels = {'rel_l2_error_y': 'State y'}

fig, ax = plt.subplots(figsize=(10,6))

epochs = np.arange(len(df)) * 500  # Each row represents 500 epochs
for key, name in labels.items():
    errs = df[key].values
    ax.semilogy(epochs[::sampling_interval],
                errs[::sampling_interval],
                lw=2, label=name)

# Annotate the final point with a short arrow
x_last = epochs[-1]
y_last = df[key].iloc[-1]
color = ax.lines[0].get_color()

mantissa, exponent = f"{y_last:.2e}".split('e')
scientific_notation = f"${mantissa} \\times 10^{{{int(exponent)}}}$"

ax.scatter([x_last], [y_last], c='k', s=30, zorder=5)
ax.annotate(
    scientific_notation,
    xy=(x_last, y_last),
    xytext=(x_last + 200, y_last),
    textcoords='data',
    va='center', ha='left',
    fontsize=14,
    color=color,
    bbox=dict(boxstyle='round,pad=0.3', fc='white', ec=color, alpha=0.8),
    arrowprops=dict(
        arrowstyle='-|>',
        lw=1,
        color=color,
        shrinkA=0,
        shrinkB=0,
        mutation_scale=10
    )
)

ax.set_xlabel('Epoch', fontsize=16)
ax.set_ylabel(r'Relative $L^2$-error', fontsize=16)
ax.tick_params(axis='both', labelsize=14)
ax.legend(fontsize=18, loc='upper right')

ax.set_xlim(0, epochs[-1] + 0.1*epochs[-1])
ticks = epochs[::4]
ax.set_xticks(ticks)

plt.tight_layout()
plt.savefig(os.path.join(figs_dir, 'example1_Forward.pdf'), bbox_inches='tight')
plt.show()
```

## Example 4: Complex domain


```python
sampling_interval = 1
loss_csv_path = "data/paper_curves/example4/loss.csv"


df = pd.read_csv(loss_csv_path)

# 1. Upper-level objective J_u_loss
fig, ax = plt.subplots(figsize=(10,6))

epochs = df['epoch'].values
j_u_values = df['J_u_loss'].values

ax.plot(epochs[::sampling_interval],
        j_u_values[::sampling_interval],
        lw=2, label='Upper-level loss $L$')

ax.set_xlabel('Epoch', fontsize=16)
ax.set_ylabel('Loss', fontsize=16)
ax.tick_params(axis='both', labelsize=14)
ax.legend(fontsize=18, loc='upper right')

plt.tight_layout()
plt.savefig(os.path.join(figs_dir, 'example3_Upper.pdf'), bbox_inches='tight')
plt.show()

# 2. Lower-level energy functional J_l_loss
fig, ax = plt.subplots(figsize=(10,6))

j_l_values = df['J_l_loss'].values

ax.plot(epochs[::sampling_interval],
        j_l_values[::sampling_interval],
        lw=2, label='Lower-level loss $\\ell$', color='#ff7f0e')

ax.set_xlabel('Epoch', fontsize=16)
ax.set_ylabel('Loss', fontsize=16)
ax.tick_params(axis='both', labelsize=14)
ax.legend(fontsize=18, loc='upper right')

plt.tight_layout()
plt.savefig(os.path.join(figs_dir, 'example3_lower.pdf'), bbox_inches='tight')
plt.show()
```

## Example 6: Non-symmetric EVI


```python
sampling_interval = 5
loss_csv_path = "data/paper_curves/example6/loss.csv"


df = pd.read_csv(loss_csv_path)

# 1. Upper-level objective J_u_loss
fig, ax = plt.subplots(figsize=(10,6))

epochs = df['epoch'].values
j_u_values = df['J_u_loss'].values

ax.plot(epochs[::sampling_interval],
        j_u_values[::sampling_interval],
        lw=2, label='Upper-level loss $L$')

ax.set_xlabel('Epoch', fontsize=16)
ax.set_ylabel('Loss', fontsize=16)
ax.tick_params(axis='both', labelsize=14)
ax.legend(fontsize=18, loc='upper right', bbox_to_anchor=(1, 0.9))

plt.tight_layout()
plt.savefig(os.path.join(figs_dir, 'example5_Upper.pdf'), bbox_inches='tight')
plt.show()

# 2. Lower-level energy functional J_l_loss
fig, ax = plt.subplots(figsize=(10,6))

j_l_values = df['J_l_loss'].values

ax.semilogy(epochs[::sampling_interval],
            j_l_values[::sampling_interval],
            lw=2, label='Lower-level loss $\\ell$', color='#ff7f0e')

ax.set_xlabel('Epoch', fontsize=16)
ax.set_ylabel('Loss (log scale)', fontsize=16)
ax.tick_params(axis='both', labelsize=14)
ax.legend(fontsize=18, loc='upper right')
ax.grid(True, which='both', ls='--', lw=0.5)

plt.tight_layout()
plt.savefig(os.path.join(figs_dir, 'example5_lower.pdf'), bbox_inches='tight')
plt.show()
```
